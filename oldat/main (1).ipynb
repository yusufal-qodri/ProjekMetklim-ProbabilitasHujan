```python
import pandas as pd

# datetime
dataCH['valid_time'] = pd.to_datetime(dataCH['valid_time'])

# ubah tp ke mm kalau belum
dataCH['tp'] = dataCH['tp'] * 1000

# kategori hujan
dataCH['Hujan'] = (dataCH['tp'] > 0.5).astype(int)

# ambil tahun dan bulan
dataCH['Year'] = dataCH['valid_time'].dt.year
dataCH['Month'] = dataCH['valid_time'].dt.month

# fungsi musim
def musim(bulan):
    if bulan in [1,2,3]:
        return 'JFM'
    elif bulan in [4,5,6]:
        return 'AMJ'
    elif bulan in [7,8,9]:
        return 'JAS'
    else:
        return 'OND'

# kolom musim
dataCH['Season'] = dataCH['Month'].apply(musim)

# hitung persentase hujan
hasilCH = (
    dataCH.groupby(['Year', 'Season'])['Hujan']
    .mean()
    .mul(100)
    .reset_index()
)

# pivot
hasilCH = hasilCH.pivot(
    index='Year',
    columns='Season',
    values='Hujan'
).reset_index()

# rename kolom
hasilCH = hasilCH.rename(columns={
    'JFM': 'CH_JFM',
    'AMJ': 'CH_AMJ',
    'JAS': 'CH_JAS',
    'OND': 'CH_OND'
})

print(hasilCH)
```

```python
gabung = pd.merge(
    hasilCH,
    dataONI,
    on='Year'
)
gabung
```

```python
corr_focus = gabung[
    ['CH_JFM', 'CH_AMJ', 'CH_JAS', 'CH_OND',
     'JFM', 'AMJ', 'JAS', 'OND']
].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr_focus,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Korelasi ONI dan Probabilitas Hujan')
plt.show()
```

```python
import pandas as pd

# datetime
dataCH['valid_time'] = pd.to_datetime(dataCH['valid_time'])

# ubah mm
dataCH['tp'] = dataCH['tp'] * 1000

# kategori hujan
dataCH['Hujan'] = (dataCH['tp'] > 0.5).astype(int)

# tahun dan bulan
dataCH['Year'] = dataCH['valid_time'].dt.year
dataCH['Month'] = dataCH['valid_time'].dt.month
```

```python
def musim(bulan):

    if bulan in [1,2,3]:
        return 'JFM'

    elif bulan in [4,5,6]:
        return 'AMJ'

    elif bulan in [7,8,9]:
        return 'JAS'

    else:
        return 'OND'

dataCH['Season'] = dataCH['Month'].apply(musim)
```

```python
spasial = (
    dataCH.groupby(
        ['Year','Season','latitude','longitude']
    )['Hujan']
    .mean()
    .mul(100)
    .reset_index()
)
```

```python
kombinasi = (
    spasial[['Year','Season']]
    .drop_duplicates()
    .values
    .tolist()
)
```

```python
from shapely.geometry import Point
import matplotlib.pyplot as plt
import geopandas as gpd
import numpy as np

# =====================================
# LOOP FRAME
# =====================================

for frame in range(0, len(kombinasi), 20):

    # =====================================
    # URUTAN MUSIM
    # =====================================

    urutan_musim = ['JFM', 'AMJ', 'JAS', 'OND']

    subset_kombinasi = kombinasi[frame:frame+20]

    subset_kombinasi = sorted(
        subset_kombinasi,
        key=lambda x: (
            x[0],
            urutan_musim.index(x[1])
        )
    )

    # =====================================
    # FIGURE
    # =====================================

    fig, axes = plt.subplots(
        4, 5,
        figsize=(20,16)
    )

    axes = axes.flatten()

    # =====================================
    # LOOP SUBPLOT
    # =====================================

    for i, (tahun, musim) in enumerate(subset_kombinasi):

        ax = axes[i]

        # =====================================
        # SUBSET DATA
        # =====================================

        subset = spasial[
            (spasial['Year'] == tahun) &
            (spasial['Season'] == musim)
        ]

        # =====================================
        # PIVOT GRID
        # =====================================

        pivot = subset.pivot(
            index='latitude',
            columns='longitude',
            values='Hujan'
        )

        # urut latitude biar tidak terbalik
        pivot = pivot.sort_index()

        # =====================================
        # GRID
        # =====================================

        lon = pivot.columns.values
        lat = pivot.index.values

        Lon, Lat = np.meshgrid(
            lon,
            lat
        )

        Z = pivot.values

        # =====================================
        # CONTOURF SPASIAL
        # =====================================

        im = ax.contourf(
            Lon,
            Lat,
            Z,
            levels=15,
            cmap='RdYlBu_r',
            extend='both'
        )

        # =====================================
        # SHAPEFILE
        # =====================================

        indo.plot(
            ax=ax,
            color='none',
            edgecolor='black',
            linewidth=0.5
        )

        # =====================================
        # HILANGKAN RONGGA PUTIH
        # =====================================

        ax.set_xlim(
            lon.min(),
            lon.max()
        )

        ax.set_ylim(
            lat.min(),
            lat.max()
        )

        # hilangkan margin putih
        ax.margins(0)

        # =====================================
        # RASIO PETA
        # =====================================

        ax.set_aspect('equal')

        # =====================================
        # JUDUL
        # =====================================

        ax.set_title(
            f'{tahun} - {musim}',
            fontsize=10
        )

        # =====================================
        # LABEL
        # =====================================

        ax.set_xlabel('')
        ax.set_ylabel('')

        # =====================================
        # TICKS
        # =====================================

        ax.tick_params(
            labelsize=7
        )

    # =====================================
    # HAPUS SUBPLOT KOSONG
    # =====================================

    for j in range(i+1, len(axes)):
        fig.delaxes(axes[j])

    # =====================================
    # LAYOUT SUBPLOT
    # =====================================

    fig.subplots_adjust(
        left=0.04,
        right=0.88,
        top=0.95,
        bottom=0.05,
        wspace=0.10,
        hspace=0.04
    )

    # =====================================
    # AXIS KHUSUS COLORBAR
    # =====================================

    cax = fig.add_axes([
        0.90,   # posisi horizontal
        0.15,   # posisi vertikal
        0.015,  # lebar
        0.70    # tinggi
    ])

    # =====================================
    # COLORBAR
    # =====================================

    cbar = fig.colorbar(
        im,
        cax=cax
    )

    cbar.set_label(
        'Probabilitas Hujan (%)',
        fontsize=11
    )

    # =====================================
    # SAVE
    # =====================================

    plt.savefig(
        f'Frame_{frame//20 + 1}.png',
        dpi=300,
        bbox_inches='tight'
    )

    # =====================================
    # SHOW
    # =====================================

    plt.show()
```

```python
# =========================
# REGRESI ONI vs
# RATA-RATA PROBABILITAS HUJAN
# =========================

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

import pandas as pd
import matplotlib.pyplot as plt

# =====================================
# STYLE PLOT
# =====================================

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 12

# =====================================
# RATA-RATA SPASIAL
# =====================================

rata_hujan = (
    spasial.groupby(['Year', 'Season'])['Hujan']
    .mean()
    .reset_index()
)

# =====================================
# UBAH ONI MENJADI FORMAT LONG
# =====================================

oni_long = dataONI.melt(
    id_vars='Year',
    value_vars=['JFM', 'AMJ', 'JAS', 'OND'],
    var_name='Season',
    value_name='ONI'
)

# =====================================
# MERGE DATA
# =====================================

gabung_regresi = pd.merge(
    rata_hujan,
    oni_long,
    on=['Year', 'Season']
)

# =====================================
# LIST MUSIM
# =====================================

musim_list = ['JFM', 'AMJ', 'JAS', 'OND']

# =====================================
# LOOP REGRESI
# =====================================

for musim in musim_list:

    # =====================================
    # FILTER DATA
    # =====================================

    subset = gabung_regresi[
        gabung_regresi['Season'] == musim
    ]

    # =====================================
    # X dan Y
    # =====================================

    X = subset[['ONI']]
    y = subset['Hujan']

    # =====================================
    # MODEL REGRESI
    # =====================================

    model = LinearRegression()
    model.fit(X, y)

    # Prediksi
    y_pred = model.predict(X)

    # =====================================
    # PARAMETER REGRESI
    # =====================================

    r2 = r2_score(y, y_pred)

    a = model.coef_[0]
    b = model.intercept_

    persamaan = f'y = {a:.3f}x + {b:.3f}'

    # =====================================
    # PLOT
    # =====================================

    plt.figure(figsize=(8,6))

    # Scatter
    plt.scatter(
        X,
        y,
        color='black',
        s=45
    )

    # Garis regresi
    plt.plot(
        X,
        y_pred,
        color='black',
        linewidth=1.5
    )

    # =====================================
    # INFORMASI
    # =====================================

    plt.text(
        0.05,
        0.95,
        f'$R^2$ = {r2:.3f}',
        transform=plt.gca().transAxes,
        fontsize=11,
        verticalalignment='top'
    )

    plt.text(
        0.05,
        0.90,
        persamaan,
        transform=plt.gca().transAxes,
        fontsize=11,
        verticalalignment='top'
    )

    # =====================================
    # LABEL
    # =====================================

    plt.title(
        f'Regresi ONI dan Probabilitas Hujan {musim}'
    )

    plt.xlabel(
        f'ONI {musim}'
    )

    plt.ylabel(
        'Probabilitas Hujan (%)'
    )

    # =====================================
    # GRID
    # =====================================

    plt.grid(
        linestyle='--',
        linewidth=0.5,
        alpha=0.5
    )

    plt.tight_layout()
    plt.show()

    # =====================================
    # OUTPUT
    # =====================================

    print(f'=== {musim} ===')
    print(f'Persamaan Regresi : {persamaan}')
    print(f'R²                : {r2:.3f}')
    print(f'Kemiringan (a)    : {a:.3f}')
    print(f'Intercept (b)     : {b:.3f}')
    print()
```
